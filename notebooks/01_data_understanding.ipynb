```python
from pathlib import Path
import pandas as pd

# Project root
PROJECT_ROOT = Path.cwd().parent

# Data directories
RAW_DATA_DIR = PROJECT_ROOT / "data" / "raw"
PROCESSED_DATA_DIR = PROJECT_ROOT / "data" / "processed"

print("Project root:", PROJECT_ROOT)
print("Raw data directory:", RAW_DATA_DIR)
print("Processed data directory:", PROCESSED_DATA_DIR)
```

```python
# Discover available data files without assuming their names or format

data_files = []

for directory in [RAW_DATA_DIR, PROCESSED_DATA_DIR]:
    if directory.exists():
        for path in directory.rglob("*"):
            if path.is_file():
                data_files.append(path)

if data_files:
    for path in data_files:
        print(path.relative_to(PROJECT_ROOT))
else:
    print("No dataset files found yet.")
```

```python
def load_tabular_file(path):
    """Load a supported tabular dataset."""
    suffix = path.suffix.lower()

    if suffix == ".csv":
        return pd.read_csv(path)
    elif suffix in [".parquet", ".pq"]:
        return pd.read_parquet(path)
    else:
        raise ValueError(f"Unsupported file format: {suffix}")


# Example: automatically inspect the first CSV/Parquet file when available
tabular_files = [
    p for p in data_files
    if p.suffix.lower() in [".csv", ".parquet", ".pq"]
]

if tabular_files:
    dataset_path = tabular_files[0]
    df = load_tabular_file(dataset_path)

    print("Loaded:", dataset_path)
    print("Shape:", df.shape)
    display(df.head())
else:
    print("No CSV/Parquet dataset available yet.")
    
```
